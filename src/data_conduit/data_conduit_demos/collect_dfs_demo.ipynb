```python
from pathlib import Path
from pprint import pprint

from data_conduit.harptools import read_harp_bin
from data_conduit.io import collect_dfs, read_csv
from data_conduit.utils import starts_with
```

```python
# collect_dfs parameter setup (edit these)
cwd = Path.cwd().resolve()
repo_root = next((p for p in [cwd, *cwd.parents] if (p / "device.yml").is_file()), cwd)

base_path = repo_root / "Misc" / "2025-10-14T18-24-42"

# Reader registry
readers = {
    ".bin": read_harp_bin,
    # ".csv": read_csv,
}

# Per-extension kwargs passed to each reader
reader_kwargs = {
    ".bin": {"harp_device_yaml_path": str(repo_root / "device.yml")},
    # ".csv": {"index_col": 0},
}

# collect_dfs behavior toggles
keep_empty = False
flatten = False
separator = ":"
verbose = False

# Optional level selectors (set to None if not needed)
l0_selector = starts_with("Behavior")
# l1_selector = None
# l2_selector = None

print("repo_root:", repo_root)
print("base_path:", base_path)
print("base exists:", base_path.is_dir())
```

```python
# collect_dfs call using the parameters above
dfs_dict = collect_dfs(
    base_path=base_path,
    readers=readers,
    reader_kwargs=reader_kwargs,
    keep_empty=keep_empty,
    flatten=flatten,
    separator=separator,
    verbose=verbose,
    l0_selector=l0_selector,
    # l1_selector=l1_selector,
    # l2_selector=l2_selector,
)

print("Top-level keys:", list(dfs_dict.keys())[:10])
print("n_top_level:", len(dfs_dict))
```
